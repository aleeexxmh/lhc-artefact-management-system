```python
"""
Artefact Management System - Local Heritage Community (LHC)
Module: Scripting and Programming Concepts (ACCA 5036)
Description: Console-based application to register, view, search, and update
historical artefacts using in-memory Python data structures.
"""

# Valid physical condition states defined by LHC specifications
VALID_CONDITIONS = ["good", "damage", "under repair"]


def get_non_empty_input(prompt):
    """Prompt the user repeatedly until a non-blank string is entered."""
    while True:
        value = input(prompt).strip()
        if value:
            return value
        print("Error: This field cannot be left blank. Please try again.")


def add_artefact(artefacts):
    """Register a new artefact dictionary and append it to the artefacts list."""
    print("\n--- Add New Artefact ---")
    name = get_non_empty_input("Enter artefact name: ")
    category = get_non_empty_input("Enter category (e.g., photograph, document): ")
    location = get_non_empty_input("Enter display/storage location: ")

    # Default condition upon initial registration is set to 'good'
    new_artefact = {
        "name": name,
        "category": category,
        "location": location,
        "condition": "good"
    }

    artefacts.append(new_artefact)
    print(f"Success: '{name}' has been added to the inventory with status 'good'.")


def update_condition(artefacts):
    """Locate an artefact by name and update its physical condition."""
    print("\n--- Update Artefact Condition ---")
    if not artefacts:
        print("No artefacts are currently stored in the system.")
        return

    target_name = get_non_empty_input("Enter the name of the artefact to update: ")
    found = False

    # Search for a case-insensitive exact match
    for item in artefacts:
        if item["name"].lower() == target_name.lower():
            found = True
            print(f"Current condition of '{item['name']}': {item['condition']}")

            # Validate that the new condition belongs to the permitted states
            while True:
                new_status = input(
                    "Enter new condition (good / damage / under repair): "
                ).strip().lower()
                if new_status in VALID_CONDITIONS:
                    item["condition"] = new_status
                    print(f"Success: Condition for '{item['name']}' updated to '{new_status}'.")
                    break
                print("Error: Invalid status. Please choose 'good', 'damage', or 'under repair'.")
            break

    if not found:
        print(f"Notice: No artefact named '{target_name}' was found.")


def view_artefacts(artefacts):
    """Display all registered artefacts and their full details in a structured format."""
    print("\n--- Artefact Inventory ---")
    if not artefacts:
        print("No artefacts are currently stored in the system.")
        return

    for index, item in enumerate(artefacts, start=1):
        print(
            f"{index}. Name: {item['name']} | Category: {item['category']} | "
            f"Location: {item['location']} | Condition: {item['condition']}"
        )


def search_artefact(artefacts):
    """Search for artefacts matching a user-provided name query (partial or full)."""
    print("\n--- Search Artefact ---")
    if not artefacts:
        print("No artefacts are currently stored in the system.")
        return

    search_term = get_non_empty_input("Enter artefact name to search: ").lower()
    matches = [item for item in artefacts if search_term in item["name"].lower()]

    if not matches:
        print(f"No artefacts found matching '{search_term}'.")
    else:
        print(f"\nFound {len(matches)} matching record(s):")
        for index, item in enumerate(matches, start=1):
            print(
                f"{index}. Name: {item['name']} | Category: {item['category']} | "
                f"Location: {item['location']} | Condition: {item['condition']}"
            )


def main():
    """Main entry point running the console menu loop."""
    artefacts = []  # In-memory list storing artefact dictionaries

    while True:
        print("\n==========================================")
        print("   LHC ARTEFACT MANAGEMENT SYSTEM")
        print("==========================================")
        print("1. Add New Artefact")
        print("2. Update Artefact Condition")
        print("3. View All Artefact Details")
        print("4. Search Artefact by Name")
        print("5. Exit")
        print("==========================================")

        choice = input("Select an option (1-5): ").strip()

        if choice == "1":
            add_artefact(artefacts)
        elif choice == "2":
            update_condition(artefacts)
        elif choice == "3":
            view_artefacts(artefacts)
        elif choice == "4":
            search_artefact(artefacts)
        elif choice == "5":
            print("Exiting Artefact Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 5.")


if __name__ == "__main__":
    main()
```
