```python
"""
Artefact Management System - Local Heritage Community (LHC)
Description: Console-based application using a nested list structure
(artefacts) with indexed fields [0: name, 1: category, 2: location, 3: condition].
"""


def add_artefact(artefacts):
    """Add a new artefact record [name, category, location, condition] to artefacts."""
    print("Add New Artefact")
    name = input("Enter artefact name:")
    category = input("Enter category (e.g., photograph, document):")
    location = input("Enter display/storage location:")
    # Default condition upon creation is set to 'good'
    condition = "good"
    new_artefact = [name, category, location, condition]
    artefacts.append(new_artefact)
    print("Success: Artefact has been added to the inventory.")


def update_condition(artefacts):
    """Update condition (index 3) for matching artefact name (index 0)."""
    print("Update Artefact Condition")
    name = input("Enter the name of the artefact to update:")
    for artefact in artefacts:
        if artefact[0] == name:
            condition = input("Enter new condition (e.g., good, damage, under repair):")
            artefact[3] = condition
            print("Artefact condition updated")


def view_artefacts(artefacts):
    """Display all stored artefacts using their four list indices (0 to 3)."""
    print("View All Artefact Details")
    if len(artefacts) == 0:
        print("No artefacts found in the inventory")
    else:
        for artefact in artefacts:
            print(artefact[0])
            print(artefact[1])
            print(artefact[2])
            print(artefact[3])


def search_artefact(artefacts):
    """Search for an artefact by matching name against index 0."""
    print("Search Artefact by Name")
    name = input("Enter the name of the artefact to search:")
    for artefact in artefacts:
        if artefact[0] == name:
            print(artefact[0])
            print(artefact[1])
            print(artefact[2])
            print(artefact[3])


def main():
    """Main loop presenting options 1 to 5 for the LHC Artefact Management System."""
    artefacts = []
    while True:
        print("LHC ARTEFACT MANAGEMENT SYSTEM")
        print("1. Add New Artefact")
        print("2. Update Artefact Condition")
        print("3. View All Artefact Details")
        print("4. Search Artefact by Name")
        print("5. Exit")
        choice = input("Select an option (1-5):")
        if choice == "1":
            add_artefact(artefacts)
        elif choice == "2":
            update_condition(artefacts)
        elif choice == "3":
            view_artefacts(artefacts)
        elif choice == "4":
            search_artefact(artefacts)
        elif choice == "5":
            print("Existing Artefact Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 5.")


if __name__ == "__main__":
    main()
```
